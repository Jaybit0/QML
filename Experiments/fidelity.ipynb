## Bit = 1

```python
def _precision_1_bit_1():
	q = QuantumCircuit(6)
	q.h(4)
	q.h(5)

	# add Rx learning
	rx = RXGate(theta=math.pi).control(num_ctrl_qubits=1)
	q.append(rx, [4, 0])
	q.append(rx, [4, 1])
	q.x(0)
	q.x(1)
	q.barrier()

	mcx = MCXGate(2)
	q.append(mcx, [0, 1, 2])

	ry = RYGate(theta=math.pi).control(num_ctrl_qubits=1)
	q.append(ry, [5, 0])
	q.append(ry, [5, 1])
	q.x(0)
	q.x(1)
	q.barrier()

	mcx = MCXGate(2)
	q.append(mcx, [0, 1, 3])
		
	return q
```
