# Increasing Precision

Objective: get runtime and gate depth counts for increasing precision

```python
import numpy as np
import qiskit as q
from qiskit import QuantumCircuit, QuantumRegister, AncillaRegister
from qiskit.quantum_info.operators import Operator
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex
import matplotlib as mpl
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  
import matplotlib.cm as cm
from qiskit.circuit.library import Diagonal
import qiskit.circuit.library as qlib
%matplotlib inline
from qiskit import transpile

from qiskit_aer import AerSimulator

import math

from qiskit.circuit.library import RXGate, MCXGate, RYGate
```

## Bit = 1

```python
def _precision_1_bit_1():
	q = QuantumCircuit(6)
	q.h(4)
	q.h(5)

	# add Rx learning
	rx = RXGate(theta=math.pi).control(num_ctrl_qubits=1)
	q.append(rx, [4, 0])
	q.append(rx, [4, 1])
	q.x(0)
	q.x(1)
	q.barrier()

	mcx = MCXGate(2)
	q.append(mcx, [0, 1, 2])

	ry = RYGate(theta=math.pi).control(num_ctrl_qubits=1)
	q.append(ry, [5, 0])
	q.append(ry, [5, 1])
	q.x(0)
	q.x(1)
	q.barrier()

	mcx = MCXGate(2)
	q.append(mcx, [0, 1, 3])
		
	return q
```

```python
def _precision_2_bit_1():
	q = QuantumCircuit(8)
	q.h(4)
	q.h(5)
	q.h(6)
	q.h(7)

	# add Rx learning
	rx_1 = RXGate(theta=math.pi).control(num_ctrl_qubits=1)
	rx_2 = RXGate(theta=math.pi / 2.).control(num_ctrl_qubits=1)
	q.append(rx_1, [4, 0])
	q.append(rx_1, [4, 1])

	q.append(rx_2, [5, 0])
	q.append(rx_2, [5, 1])
	q.x(0)
	q.x(1)
	q.barrier()

	mcx = MCXGate(2)
	q.append(mcx, [0, 1, 2])

	ry_1 = RYGate(theta=math.pi).control(num_ctrl_qubits=1)
	ry_2 = RYGate(theta=math.pi / 2).control(num_ctrl_qubits=1)
	q.append(ry_1, [6, 0])
	q.append(ry_1, [6, 1])
	q.append(ry_2, [7, 0])
	q.append(ry_2, [7, 1])
	q.x(0)
	q.x(1)
	q.barrier()

	mcx = MCXGate(2)
	q.append(mcx, [0, 1, 3])

	# q.draw()
		
	return q
```

```python
def _precision_3_bit_1():
	q = QuantumCircuit(10)

	for i in range(4, 10):
		q.h(i)

	# add Rx learning
	rx_1 = RXGate(theta=math.pi).control(num_ctrl_qubits=1)
	rx_2 = RXGate(theta=math.pi / 2.).control(num_ctrl_qubits=1)
	rx_3 = RXGate(theta=math.pi / 4.).control(num_ctrl_qubits=1)
	q.append(rx_1, [4, 0])
	q.append(rx_1, [4, 1])
	q.append(rx_2, [5, 0])
	q.append(rx_2, [5, 1])
	q.append(rx_3, [6, 0])
	q.append(rx_3, [6, 1])
	q.x(0)
	q.x(1)
	q.barrier()

	mcx = MCXGate(2)
	q.append(mcx, [0, 1, 2])

	ry_1 = RYGate(theta=math.pi).control(num_ctrl_qubits=1)
	ry_2 = RYGate(theta=math.pi / 2.).control(num_ctrl_qubits=1)
	ry_3 = RYGate(theta=math.pi / 4.).control(num_ctrl_qubits=1)
	q.append(ry_1, [7, 0])
	q.append(ry_1, [7, 1])
	q.append(ry_2, [8, 0])
	q.append(ry_2, [8, 1])
	q.append(ry_3, [9, 0])
	q.append(ry_3, [9, 1])
	q.x(0)
	q.x(1)
	q.barrier()

	mcx = MCXGate(2)
	q.append(mcx, [0, 1, 3])
		
	return q
```

## Bit = 2

```python
def _precision_1_bit_2():
	q = QuantumCircuit(13)
	
	for i in range(8, 13):
		q.h(i)

	# bit 1
	# add Rx learning
	rx_1 = RXGate(theta=math.pi).control(num_ctrl_qubits=1)
	q.append(rx_1, [8, 0])
	q.append(rx_1, [8, 1])
	q.x(0)
	q.x(1)

	mcx = MCXGate(2)
	q.append(mcx, [0, 1, 4])

	# add Ry learning
	ry_1 = RYGate(theta=math.pi).control(num_ctrl_qubits=1)
	q.append(ry_1, [10, 0])
	q.append(ry_1, [10, 1])
	q.x(0)
	q.x(1)
	
	mcx = MCXGate(2)
	q.append(mcx, [0, 1, 6])

	q.append(mcx, [0, 12, 2])
	q.append(mcx, [1, 12, 3])

	q.barrier()

	# bit 2
	# add Rx learning
	q.append(rx_1, [9, 2])
	q.append(rx_1, [9, 3])
	q.x(2)
	q.x(3)

	q.append(mcx, [2, 3, 5])

	# add Ry learning
	q.append(ry_1, [11, 2])
	q.append(ry_1, [11, 3])
	q.x(2)
	q.x(3)
	
	mcx = MCXGate(2)
	q.append(mcx, [2, 3, 7])
		
	return q

# _precision_1_bit_2().draw()
```
